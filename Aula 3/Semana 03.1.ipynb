Exercício 5 — Tamanho de imagem Full HD
Nível: Aplicação Objetivo: Calcular o tamanho bruto, em bytes e MB, de uma imagem
RGB sem compressão.
Enunciado
Calcule o tamanho bruto, em bytes e em MB, de uma imagem RGB de 1920 × 1080 pixels
sem compressão.
Como pensar antes de programar
Este cálculo combina três ideias: a quantidade total de pixels, o número de bytes que cada
pixel ocupa (visto no Capítulo 2) e a conversão de bytes para KB e de KB para MB usando
a base 1.024
asso O que fazer / O que acontece
## 1 — Pixels Calcule a quantidade total de pixels: largura × altura = 1920 × 1080 = 2.073.600
pixels.##
2 — **Bytes por
pixel**
**Cada pixel RGB usa 3 bytes: 1 para vermelho (R), 1 para verde (G) e 1 para
azul (B)**
3 — **Bytes totais Bytes = pixels × 3 = 2.073.600 × 3 = 6.220.800 bytes**.
4 — **KB KB = bytes ÷ 1.024 = 6.220.800 ÷ 1.024 = 6.075 KB**.
5 — **MB MB = KB ÷ 1.024 = 6.075 ÷ 1.024 ≈ 5,93 MB**.

```python
largura = 1920
altura = 1080
bytes_por_pixel = 3 # R=1 byte, G=1 byte, B=1 byte
tamanho_bytes = largura * altura * bytes_por_pixel
tamanho_kb = tamanho_bytes / 1024
tamanho_mb = tamanho_kb / 1024
print(f'Pixels totais: {largura * altura:,}')
print(f'Tamanho: {tamanho_bytes:,} bytes')
print(f'Tamanho: {tamanho_kb:,.0f} KB')
print(f'Tamanho: {tamanho_mb:.2f} MB')
```
