 Um byte vale 8 bits.

RGB siginfica Red, Green e blue, seu tamanho em pixels se mede por altura * largura* 3

```python
qtde_imagens = 50_000
largura = 1920
altura = 1080
bytes_por_pixel = 3
bytes_por_imagem = largura * altura * bytes_por_pixel
total_bytes = bytes_por_imagem * qtde_imagens
total_gb = total_bytes / 1024**3
mb_unit = bytes_por_imagem / 1024 / 1024

print(f'Por imagem: {bytes_por_imagem:,} bytes ({mb_unit:.1f} MB)')
print(f'Total bruto: {total_bytes:,} bytes')
print(f'Total bruto: {total_gb:.2f} GB')
#Por imagem: 6,220,800 bytes (5.9 MB)
#Total bruto: 311,040,000,000 bytes
#Total bruto: 289.68 GB
#Cálculo 2 — com JPEG de 500 KB por imagem:
kb_por_imagem_jpeg = 500
total_kb_jpeg = kb_por_imagem_jpeg * qtde_imagens
total_gb_jpeg = total_kb_jpeg / 1024**2
print(f'Total JPEG: {total_kb_jpeg:,} KB')
print(f'Total JPEG: {total_gb_jpeg:.2f} GB')
reducao = total_bytes / (total_kb_jpeg * 1024)
print(f'Redução: {reducao:.1f}x menos espaço com JPEG')
```
