### Importing Data from Google Sheets

To access a Google Sheet from Colab, you'll typically need to:
1.  **Authenticate** your Colab notebook with your Google account.
2.  **Share** the Google Sheet with the service account that Colab uses, or make it publicly accessible (less secure for sensitive data).
3.  **Get the Sheet ID** from the URL of your Google Sheet.

Let's start with the authentication and then load the data.

```python
# Install necessary libraries if you haven't already
!pip install --upgrade gspread pandas gspread-dataframe

# Import libraries
import gspread
from google.colab import auth
from gspread_dataframe import get_dataframe, set_dataframe
import pandas as pd
```

### Step 1: Authenticate with Google Drive

Run the following cell to authenticate your Colab session with your Google account. This will open a new window or tab for you to log in and grant permissions.

```python
# Authenticate Google Colab to access Google Drive
auth.authenticate_user()

gc = gspread.service_account()
print('Authentication complete.')
```

### Step 2: Load data from Google Sheet

Before running the next cell:
1.  **Open your Google Sheet.**
2.  **Get the spreadsheet ID** from the URL. It's the long string of characters between `/d/` and `/edit` (e.g., `1Bxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx`).
3.  **Make sure the sheet is accessible.** The easiest way for demonstration is to change the 'Share' settings to 'Anyone with the link can view'. Alternatively, you can share it specifically with the email address of the service account you authenticate with (usually displayed during the authentication step, or found in `gc.auth.service_account_email`).

Replace `'YOUR_SPREADSHEET_ID_HERE'` with your actual spreadsheet ID.

```python
print("olá mundo")
```

```python
# Replace 'YOUR_SPREADSHEET_ID_HERE' with your actual Google Sheet ID
spreadsheet_id = 'YOUR_SPREADSHEET_ID_HERE'

try:
    # Open the spreadsheet by ID
    worksheet = gc.open_by_key(spreadsheet_id).worksheet('Sheet1') # Replace 'Sheet1' with your sheet name if different

    # Read the data into a Pandas DataFrame
    df = get_dataframe(worksheet)

    print('Data loaded successfully!')
    display(df.head())
except Exception as e:
    print(f"An error occurred: {e}")
    print("Please ensure your spreadsheet ID is correct and the sheet is properly shared.")
```
