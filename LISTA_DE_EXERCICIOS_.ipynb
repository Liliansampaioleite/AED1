
Exercício 1 — Maioridade
Nível: Aquecimento Objetivo: Usar if/else para decidir entre dois caminhos a partir de
uma comparação simples (idade >= 18).
Enunciado
Peça a idade da pessoa e exiba se ela é maior de idade (18 anos ou mais) ou menor de
idade.
Como pensar antes de programar
A regra que você já conhece — "maior de idade é ter 18 anos ou mais" — é uma condição:
em Python, toda condição produz True ou False. Aqui existem exatamente dois caminhos
possíveis: a pessoa é maior de idade, ou não é. Quando há exatamente dois caminhos
mutuamente exclusivos, a estrutura correta é if/else. O bloco do if executa quando a
condição é True; o bloco do else executa quando é False — nunca os dois ao mesmo tempo.

```python
idade = int(input("Digite a sua idade: "))
if idade >= 18:
  print("Você é maior de idade")
else:
  print("Você é menor de idade")
```

mutuamente exclusivas (positivo, negativo ou zero).
Enunciado

Como pensar antes de programar
Diferente do exercício anterior, aqui existem três possibilidades, e elas não podem ocorrer
ao mesmo tempo para o mesmo número. Quando as situações são alternativas exclusivas
como essa, a estrutura correta é if/elif/else. A ordem importa: se o número for maior
que zero, é positivo. Se não for — já sabemos que não é maior que zero —, pode ser
negativo ou zero; testamos então se é menor que zero, e se ainda não for nenhum dos dois, só pod ser zero
